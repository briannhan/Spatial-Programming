```python
import arcpy
import os
from pathlib import Path

cwd = os.getcwd()
arcpy.env.workspace = Path(cwd)
arcpy.overwriteOutput = True
```

```python
japanInlandWaterFilename = "Japan_Inland_Water/Japan_Inland_Water.shp"
desc_JapanInlandWater = arcpy.Describe(japanInlandWaterFilename)
```

```python
desc_JapanInlandWater
```

```python
japanRivers = arcpy.analysis.Select(japanInlandWaterFilename, "Japan_Inland_Water/japanRivers.shp", "hyt = 1")
```

```python
japanRivers
```

```python
japanLandFilename = r"Japan_Land/Japan_Land.tif"
japanLand = arcpy.sa.Raster(japanLandFilename)
japanLand
```

```python
japanUrban = arcpy.sa.ExtractByAttributes(japanLandFilename, "VALUE = 70")
japanUrban.save("Japan_Land/Japan_Urban.tif")
```
