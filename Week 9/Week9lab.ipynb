```python
import arcpy
from pathlib import Path
import os
```

# Question 2

```python
cwd = Path(os.getcwd())
arcpy.env.workspace = cwd
arcpy.overwriteOutput = True  # Error 1: not setting overwriteOutput to True
```

```python
bangladeshPath = cwd/"Bangladesh_Hospitals/Bangladesh_Hospitals.shp"
desc = arcpy.Describe(bangladeshPath)
```

```python
# The code continues if the feature class is a shapefile
if desc.dataType == "ShapeFile":
    
    # These two variables will be used to construct SQL queries 
    amenity1 = "hospital"
    amenity2 = "pharmacy"
    
    # Create a new feature class that contains only pharmacies from the original
    pharmFeature = cwd/"Bangladesh_Hospitals"/"Bangladesh_Pharmacies.shp"
    whereClause = "amenity = '{0}'".format(amenity1)
    arcpy.analysis.Select(bangladeshPath,pharmFeature,whereClause)  # Error 2: misspelling "pharmFeature" as "pharmfeature"
    
    # Same again but for the point locations of hospitals only
    hospFeature = "Bangladesh_Hosp_Only"
    whereClause = "amenity = '{0}'".format(amenity2)  # Error 3: having 2 equal signs in a SQL query statement
    arcpy.analysis.Select(bangladeshPath,hospFeature,whereClause)
# The else condition catches the application of a feature class that does
# not contain a shapefile
else
    print("Please check the data type of the downloaded folder and try again.")
```

# Question 3

```python
totalRows = arcpy.management.GetCount(bangladeshPath)
totalRowsNumber = int(totalRows[0])
emptySourceLayer, emptySourceRows = arcpy.management.SelectLayerByAttribute(bangladeshPath, where_clause="source = ''")
emptySourcePercent = 100 * int(emptySourceRows)/totalRowsNumber

sourceSpecifiedLayer, sourceSpecifiedRows = arcpy.management.SelectLayerByAttribute(bangladeshPath, where_clause="source = ''",
                                                                                    invert_where_clause="INVERT")
sourceSpecifiedPercent = 100 * int(sourceSpecifiedRows)/totalRowsNumber
print("{0:.2f}% of hospitals and pharmacies in Bangladesh have a source while {1:.2f}% of hospitals and pharmacies in Bangladesh do not have a source".format(sourceSpecifiedPercent, emptySourcePercent))
```
